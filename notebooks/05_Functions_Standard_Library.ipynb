# Built_in Functions and Standard Library

## 1. Built-in Functions

### 1.1 int()

It returns an integer

```python
int(45.6666)
```

```python
int("45")
```

### 1.2 'bool()'

it returns 'True' or 'False'

```python
# If it takes a number that is not zero. it will return True

bool(45)
```

```python
bool(0) # If it takes a zero, then it will return False
```

```python
bool("Aaa")
```

### 1.3 'float()'

It returns the flaoting-point type values.

```python
float(455)
```

```python
float("4")
```

### 1.4 'str()'

it returns the string type values 

```python
str(455)
```

```python
str(45.0)
```

```python
str(True)
```

### 1.5 'str.lower()' method

It returns the lower case of the given string

```python
string1 = "Mike Tyson"

string1
```

```python
string2 = string1.lower()

string2
```
