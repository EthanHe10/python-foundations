## 2. Object References

Python provides the 'is' and 'is not' operators for compaering references.

```python
name1 = "Smith"
name2 = "John"
```

```python
name1 is name2  # Checking if it's the same
```

```python
name2 = name1

name2  # name2 is alias of name1 now.
```

```python
name1 is name2
```

```python
name3 = "Mike"

name1 is not name3
```

```python
name3 is not None
```
